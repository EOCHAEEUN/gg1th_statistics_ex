### 단일표본 T 검정 

```python
with open("./data2/성인여성_키_데이터 (1).txt", "r",encoding="cp949") as f:
    data = f.read().split('\n')
    data = list (map(float,data))
    print(data)
    print(len(data))
```

```python
data [:3]
```

```python
# 데이터의 평균과 표준편차 구하기 

from scipy import stats
import numpy as np

mean_value = np.mean(data)
std_dev = np.std(data, ddof=0)  # 모집단 표준편차 (기본설정)

mean_value, std_dev
```

```python
#가설 설정 
#귀무가설 : 표본 데이터의 평균은 163과 같다 -> 대립가설: 다르다     
```

```python
result = stats.kstest(
    data,
    stats.norm.cdf,
    args=(mean_value, std_dev)
)
print(result)
```

```python
stats.shapiro(data)
```

### 검정하기 
- 귀무가설 : 표본 데이터의 평균은 163과 같다  ->대립가설 : 다르다

```python
#단일 표본 t-검정 수행
print(stats.ttest_1samp(data, 163))
```

### t-통계량
- statistic = -1.6747 : t-값이 음수인 것은 표본 평균이 163보다 작다는 것을 의미
- 절댓값이 클수록 귀무가설을 기각할 가능성이 커짐


### 독립표본 t 검정
- 두 그룹 A,B 반의 점수차이가 있는지 판단하기
- 가설: 귀무가설 A와 B반의 평균은 같다 -> 대립 ; 다르다  

```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv("datas2/반별_점수_type1.csv", encoding = "euc-kr")
# df1 = pd.read_csv("./data2/반별_점수_type2.csv")
df1.head()
```

```python

#A반 , B반 데이터 분리 
# A반의 데이터만 추출
group_A = df1['점수'].loc[df1['반'] == 'A'].values  # loc를 사용하여 '반'이 'A'인 행의 '점수' 열을 선택하고, values 속성을 사용하여 numpy 배열로 변환
#loc -> 위치를 사용해서 슬라이싱 or 배열 
# B반의 데이터만 추출
group_B = df1['점수'].loc[df1['반'] == 'B'].values

print(group_A)
print(group_B)
```

```python
import numpy as np

# A반의 평균과 표준 편차
mean_A = np.mean(group_A)
std_A = np.std(group_A, ddof=1)  # 표본 표준 편차 (ddof=1)

# B반의 평균과 표준 편차
mean_B = np.mean(group_B)
std_B = np.std(group_B, ddof=1)  # 표본 표준 편차 (ddof=1)

# 결과 출력
print(f"A반 평균: {mean_A:.2f}, 표준 편차: {std_A:.2f}")
print(f"B반 평균: {mean_B:.2f}, 표준 편차: {std_B:.2f}")
```

```python
# group_A와 group_B의 평균과 표준편차를 사용하여 정규성 검정 수행
# 'norm': 비교 대상이 되는 이론적인 분포를 지정 ('norm'은 정규분포) -> SciPy 1.18.0 버전부터 stats.norm.cdf 로 변경
print(stats.kstest(group_A, stats.norm.cdf, args=(np.mean(group_A), np.std(group_A, ddof=0))))
print(stats.kstest(group_B, stats.norm.cdf, args=(np.mean(group_B), np.std(group_B, ddof=0))))
```

```python
stat, p = stats.shapiro(group_A) 
print(f"group_A의 stat: {stat}, p: {p}")
stat, p = stats.shapiro(group_B) 
print(f"group_B의 stat: {stat}, p: {p}")
```
