## 쌍체 표본 t검정
 - 다이어트 전과 후의 차이가 유의한지 

```python
import pandas as pd
```

```python
df = pd.read_csv("datas2/다이어트약_효과검증.csv", encoding = "utf-8'")
```

```python

df.columns
```

```python
#A그룹 
before = df['다이어트전(kg)']
#B그룹 
after = df['다이어트후(kg)']
```

```python
from scipy import stats
stats.shapiro(after-before)    # Shapiro-Wilk 검정 정규성 검정
```

```python
stats.normaltest(after-before)  # D'Agostino-Pearson 정규성 검정 
```

```python
stats.ttest_rel(before, after)  #쌍체표본 t검정
```

```python
import numpy as np
np.mean(before), np.mean(after)  #평균값 확인
```

```python
np.mean(before) - np.mean(after)  #평균값 확인
```

```python
#전후 차이 계산(개인별 체중차이)
diff = before - after

#체중 차이 평균계산 (평균 체중감소량)
mean_diff = np.mean(diff)

#체중 차이의 표준편차 (감소량의 표준편차)
std_diff = np.std(diff, ddof=1)

#체중 차이의 표준오차 (감소량의 표준오차=표본에서 구한 평균체중 감소량의 불확실성을 나타냄)
se_diff = std_diff / np.sqrt(len(diff))

#t-통계량 계산 (평균체중 감소량이 표준오차보다 몇 배 큰지를 나타냄)
t_statistic = mean_diff / se_diff
t_statistic
```

```python
#계산 결과 출력
print("평균 체중 감소량: {:.2f} kg".format(mean_diff))
print("체중 감소량의 표준편차: {:.2f} kg".format(std_diff))
print("체중 감소량의 표준오차: {:.2f} kg".format(se_diff))
print("t-통계량: {:.2f}".format(t_statistic))
```
